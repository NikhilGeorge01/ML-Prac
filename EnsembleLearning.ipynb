```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
data = load_iris()
```

```python
dt = DecisionTreeClassifier(max_depth=3, random_state=40)
rf = RandomForestClassifier(n_estimators=3, random_state=40)
X = data.data
y = data.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state=40)
```

```python
rf.fit(X_train,y_train)
```

```python
dt.fit(X_train,y_train)
```

```python
rf_pred = rf.predict(X_test)
dt_pred = dt.predict(X_test)
```

```python
print("FOR DECISION TREE")
print(f"Accuracy is {accuracy_score(y_test,dt_pred)} Precision is {precision_score(y_test,dt_pred, average='macro')} Recall is {recall_score(y_test,dt_pred, average='macro')} F1 score is {f1_score(y_test,dt_pred, average='macro')}")
```

```python
print("FOR RANDOM FOREST")
print(f"Accuracy is {accuracy_score(y_test,rf_pred)} Precision is {precision_score(y_test,rf_pred, average='macro')} Recall is {recall_score(y_test,rf_pred, average='macro')} F1 score is {f1_score(y_test,rf_pred, average='macro')}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
log_clf = LogisticRegression(max_iter=1000)
rf_clf = RandomForestClassifier(n_estimators=100, random_state=40)
knn_clf = KNeighborsClassifier(n_neighbors=5)
from sklearn.ensemble import VotingClassifier
```

```python
voting_hard = VotingClassifier(
    estimators=[('lr', log_clf), ('rf', rf_clf), ('knn', knn_clf)],
    voting='hard'
)

voting_hard.fit(X_train, y_train)
y_pred = voting_hard.predict(X_test)
print("Hard Voting Accuracy:", accuracy_score(y_test, y_pred))
```

```python
voting_soft = VotingClassifier(
    estimators=[('lr', log_clf), ('rf', rf_clf), ('knn', knn_clf)],
    voting='soft'
)

voting_soft.fit(X_train, y_train)
y_pred = voting_soft.predict(X_test)
print("Soft Voting Accuracy:", accuracy_score(y_test, y_pred))
```

```python
voting_weighted = VotingClassifier(
    estimators=[('lr', log_clf), ('rf', rf_clf), ('knn', knn_clf)],
    voting='soft',
    weights=[1, 3, 2]  # higher weight to Random Forest
)

voting_weighted.fit(X_train, y_train)
y_pred = voting_weighted.predict(X_test)
print("Weighted Soft Voting Accuracy:", accuracy_score(y_test, y_pred))
```

```python
from sklearn.datasets import load_diabetes
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

data_reg = load_diabetes()
X_reg = data_reg.data
y_reg = data_reg.target

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.2, random_state=40)

rf_reg = RandomForestRegressor(n_estimators=100, random_state=40, oob_score=True)
rf_reg.fit(X_train_reg, y_train_reg)

print("OOB Score:", rf_reg.oob_score_)

y_pred_reg = rf_reg.predict(X_test_reg)
print("MSE on Test Set:", mean_squared_error(y_test_reg, y_pred_reg))
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

data_bc = load_breast_cancer()
X_bc = data_bc.data
y_bc = data_bc.target

X_train_bc, X_test_bc, y_train_bc, y_test_bc = train_test_split(X_bc, y_bc, test_size=0.2, random_state=40)

adaboost = AdaBoostClassifier(n_estimators=100, random_state=40)
adaboost.fit(X_train_bc, y_train_bc)

y_pred_bc = adaboost.predict(X_test_bc)
print("AdaBoost Accuracy:", accuracy_score(y_test_bc, y_pred_bc))
```
